## Project Title

**Exploring the Relationship Between Health Outcomes and Development Indicators: A Global Analysis**

## Problem Statement

Health outcomes vary considerably across countries and over time. Differences in economic development, healthcare investment, healthcare capacity, and access to essential services may be associated with these variations.

This project uses World Bank World Development Indicators (WDI) data to examine relationships between selected socioeconomic, healthcare, and basic-service indicators and population health outcomes across countries and over time.

## Project Objective

The objective of this project is to analyse global health and development data to identify trends, patterns, and relationships between selected development indicators and health outcomes. The analysis will provide evidence-based insights that can support understanding of factors associated with differences in population health.

## Research Question

**How are economic development, healthcare investment, healthcare capacity, and access to sanitation associated with health outcomes across countries?**

The analysis will focus on:

* Life expectancy as the primary health outcome.
* Infant mortality as an additional health outcome.

* Access to basic sanitation as an indicator of essential services.

The analysis will examine trends over time, differences between countries and regions, and relationships between the selected development indicators and health outcomes.

### Change in Indicators, 2021–2023

The change in each indicator between 2021 and 2023 is calculated to quantify the direction and magnitude of change over the study period. This helps identify which indicators improved, declined, or remained relatively stable.

```python
change_2021_2023 = (
    df_analysis.groupby("Indicator Name")[["2021", "2023"]]
    .mean()
)

change_2021_2023["Absolute Change"] = (
    change_2021_2023["2023"] - change_2021_2023["2021"]
)

change_2021_2023["Percentage Change"] = (
    (change_2021_2023["2023"] - change_2021_2023["2021"])
    / change_2021_2023["2021"]
) * 100

change_2021_2023
```

### Interpretation

Between 2021 and 2023, most of the selected indicators showed improvement. Average GDP per capita increased by 5.17%, while life expectancy increased by 2.83%. Infant mortality decreased by 4.54%, indicating an improvement in this health outcome. Access to basic sanitation also increased by 1.69%.

However, average current health expenditure as a percentage of GDP decreased by 4.59% over the same period. Overall, the results suggest improvements in several health and socioeconomic outcomes, although health expenditure as a share of GDP declined.

### Correlation Analysis

Correlation analysis is used to examine the strength and direction of relationships between the selected health and socioeconomic indicators across countries in 2023. The analysis identifies potential associations but does not establish causal relationships.

```python
df_2023 = df_analysis.pivot(
    index=["Country Name", "Country Code"],
    columns="Indicator Name",
    values="2023"
).reset_index()

df_2023.head()
```

### Interpretation

The 2023 data was reshaped so that each country or region is represented by one row, with the five selected indicators represented as separate columns. This structure allows comparisons between indicators and supports correlation analysis across countries and regions.

```python
correlation_matrix = df_2023[
    selected_indicators
].corr()

correlation_matrix
```

### Interpretation

The correlation analysis for 2023 shows several notable relationships between the selected health and socioeconomic indicators. Life expectancy has a strong negative correlation with infant mortality (r = -0.90), indicating that countries with higher life expectancy generally tend to have lower infant mortality. Life expectancy also has a strong positive correlation with basic sanitation coverage (r = 0.83) and a moderate positive correlation with GDP per capita (r = 0.63).

Infant mortality has a strong negative correlation with sanitation coverage (r = -0.86) and a moderate negative correlation with GDP per capita (r = -0.48). GDP per capita has a moderate positive correlation with sanitation coverage (r = 0.42).

Current health expenditure as a percentage of GDP shows relatively weak correlations with the other indicators, ranging from approximately -0.29 to 0.25. This suggests that the proportion of GDP allocated to health expenditure alone does not show a strong linear association with the selected health outcomes in this dataset.

These correlations indicate associations between the indicators and do not establish causal relationships.

### Correlation Heatmap

A correlation heatmap is used to visually represent the strength and direction of relationships between the selected indicators in 2023. This makes it easier to identify the strongest positive and negative associations.

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 7))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0
)

plt.title("Correlation Between Selected Health and Socioeconomic Indicators — 2023")
plt.tight_layout()
plt.show()
```

### Interpretation

The heatmap confirms several strong relationships among the selected indicators in 2023. Life expectancy has a very strong negative correlation with infant mortality (r = -0.90) and a very strong positive correlation with basic sanitation coverage (r = 0.83). It also has a strong positive correlation with GDP per capita (r = 0.63).

Infant mortality is negatively associated with sanitation coverage (r = -0.86) and GDP per capita (r = -0.48). Current health expenditure has relatively weak correlations with the other indicators, suggesting that its percentage of GDP alone has a limited linear association with the selected health outcomes.

Overall, the heatmap highlights sanitation, economic conditions, and health outcomes as important areas for further investigation. These relationships represent associations and should not be interpreted as evidence of causation.

## Key Findings and Insights

The analysis of the five selected World Development Indicators across 2021–2023 revealed several important patterns in economic conditions, healthcare, infrastructure, and population health.

### 1. Health outcomes improved between 2021 and 2023

Average life expectancy increased from 71.47 years in 2021 to 73.49 years in 2023, representing a 2.83% increase. Over the same period, average infant mortality declined from 21.18 to 20.22 deaths per 1,000 live births, a 4.54% reduction. These changes indicate an overall improvement in the selected population health measures during the study period.

### 2. Economic conditions showed moderate improvement

Average GDP per capita increased from approximately US$16,069 in 2021 to US$16,899 in 2023, representing a 5.17% increase. However, the large variation between countries remained evident, with 2023 GDP per capita ranging from approximately US$265 in Burundi to over US$225,884 in Monaco.

### 3. Access to basic sanitation improved, but major disparities remained

Average access to at least basic sanitation services increased from 78.33% in 2021 to 79.66% in 2023. Despite this improvement, substantial differences remained. Some countries recorded 100% coverage, while Ethiopia recorded approximately 10.02%, demonstrating a major gap in access to basic sanitation services.

### 4. Sanitation showed the strongest relationship with the selected health outcomes

Sanitation coverage had a strong positive correlation with life expectancy (r = 0.83) and a strong negative correlation with infant mortality (r = -0.86). These were the strongest relationships identified in the correlation analysis. Countries with higher sanitation coverage generally tended to have higher life expectancy and lower infant mortality.

### 5. Economic prosperity was associated with better health outcomes

GDP per capita showed a moderately strong positive association with life expectancy (r = 0.63) and a moderate negative association with infant mortality (r = -0.48). This suggests that countries with higher economic resources generally experienced more favourable health outcomes, although GDP alone does not explain all differences between countries.

### 6. Healthcare expenditure showed weaker associations with health outcomes

Current health expenditure as a percentage of GDP had relatively weak correlations with life expectancy (r = 0.25) and infant mortality (r = -0.29). This suggests that the proportion of GDP allocated to current healthcare expenditure alone may not adequately explain differences in health outcomes across countries.

### 7. Large inequalities exist across countries

The country-level comparisons demonstrate substantial differences across all five indicators. In 2023, life expectancy ranged from 54.46 to 86.37 years, infant mortality ranged from 1.3 to 70.1 deaths per 1,000 live births, and sanitation coverage ranged from approximately 10% to 100%. These differences highlight persistent inequalities in health, economic conditions, and infrastructure across countries and regions.

## Recommendations

Based on the findings from the analysis, the following recommendations are proposed:

### 1. Strengthen access to basic sanitation services

The strong association between sanitation coverage and health outcomes highlights the importance of improving access to basic sanitation, particularly in countries with very low coverage. Investment in sanitation infrastructure, hygiene services, and safe waste management could support improvements in population health.

### 2. Prioritise countries with poor health outcomes

Countries recording low life expectancy and high infant mortality should receive greater attention when allocating health and development resources. Data-driven identification of countries with poorer outcomes can help policymakers target interventions where the need is greatest.

### 3. Address economic inequalities

The positive association between GDP per capita and life expectancy, together with the negative association between GDP per capita and infant mortality, suggests that broader socioeconomic conditions are important when addressing health inequalities. Policies that support economic development, employment, education, and access to essential services should therefore complement healthcare interventions.

### 4. Improve the effectiveness of healthcare investment

The relatively weak correlation between current health expenditure as a percentage of GDP and the selected health outcomes suggests that simply increasing the proportion of GDP spent on healthcare may not be sufficient. Policymakers should also consider how healthcare resources are allocated, the quality and accessibility of services, and whether spending reaches populations with the greatest needs.

### 5. Strengthen health and development data systems

The analysis identified differences in indicator coverage and missing observations across countries. Strengthening routine data collection, reporting, and data quality systems would improve the ability of policymakers and researchers to monitor health inequalities and evaluate interventions over time.

### 6. Use integrated approaches to improve population health

The relationships observed in the analysis suggest that health outcomes are connected with economic conditions and infrastructure. Public health strategies should therefore combine healthcare interventions with improvements in sanitation, socioeconomic conditions, and access to essential services rather than focusing on healthcare expenditure alone.

## Conclusion

This analysis examined five World Development Indicators across 2021–2023: current health expenditure, GDP per capita, life expectancy, infant mortality, and access to basic sanitation services. The analysis identified substantial differences in health, economic conditions, and infrastructure across countries and regions.

Between 2021 and 2023, average GDP per capita and life expectancy increased, while infant mortality declined. Access to basic sanitation also improved slightly. However, substantial inequalities remained between countries, particularly in life expectancy, infant mortality, GDP per capita, and sanitation coverage.

The correlation analysis identified the strongest relationships between sanitation and the selected health outcomes. Sanitation coverage had a strong positive association with life expectancy (r = 0.83) and a strong negative association with infant mortality (r = -0.86). GDP per capita also showed a positive association with life expectancy (r = 0.63) and a negative association with infant mortality (r = -0.48). In contrast, current health expenditure as a percentage of GDP showed weaker relationships with the selected health outcomes.

Overall, the findings suggest that population health is associated with broader socioeconomic and infrastructure conditions, particularly economic resources and access to basic sanitation. The results highlight the importance of integrated approaches to health improvement that consider healthcare, sanitation, and socioeconomic development together. However, the observed relationships are associative and should not be interpreted as evidence of causation.
